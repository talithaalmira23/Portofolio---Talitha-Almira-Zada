```python
import re

# Mendefinisikan Regex untuk mengidentifikasi pattern URL
pattern = r'http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\\(\\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+'

# Function untuk menghilangkan URL
def remove_urls(text):
    # Handle non-string values
    if not isinstance(text, str):
        return text  # Return the original value if it's not a string
    return re.sub(pattern, '', text)

# Menggunakan function pada kolom 'full_text'
column_name = 'full_text'
df[column_name] = df[column_name].apply(remove_urls)
```

**Clear Spesific Word/**

```python
#Menghilangkan "its/" pada awalan text
df['full_text'] = df['full_text'].str.replace('kata/', '')

#Test
first_10_rows = df.head(10)
print(first_10_rows)
```

**Make sure all word in alphabet**

```python
import re
import pandas as pd

# Pola regex untuk identifikasi karakter non-alphabet
pattern = r'[^a-zA-Z\s]+'

# Function untuk menghapus karakter non-alphabet
def remove_non_alphabet(text):
    return re.sub(pattern, '', text)

# Mengubah value text menjadi string
df['full_text'] = df['full_text'].apply(lambda x: str(x) if not isinstance(x, str) else x)

# Menggunakan function pada kolom 'full_text'
df['full_text'] = df['full_text'].apply(remove_non_alphabet)

#Test
first_10_rows = df.head(10)
print(first_10_rows)
```

**Remove Punctuation**

```python
import string

# Regex function untuk menghilangkan tanda baca
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

# Menggunakan function pada kolom 'full_text'
df['full_text'] = df['full_text'].apply(remove_punctuation)

#Test
first_10_rows = df.head(10)
print(first_10_rows)
```

**Remove Leading Space & Extra Whitespace**

```python
# Menghilangkan leading space
df['full_text'] = df['full_text'].str.lstrip()

# Function untuk menghilangkan extra whitespace pada text
def remove_extra_spaces(text):
    return re.sub(r'\s+', ' ', text).strip()

# Apply the function to the specified column.
df['full_text'] = df['full_text'].apply(remove_extra_spaces)

# Test
first_10_rows = df.head(10)
print(first_10_rows)
```

**Remove 1 word item**

```python
# Function untuk menghilangkan kalimat yang hanya terdiri dari 1 kata
def remove_single_word(text):
    sentences = text.split('.')
    filtered_sentences = [sentence.strip() for sentence in sentences if len(sentence.split()) > 1]
    return '. '.join(filtered_sentences)

# Apply the function to the specified column.
df['full_text'] = df['full_text'].apply(remove_single_word)

# Test
first_10_rows = df.head(10)
print(first_10_rows)
```

**Remove additional stopword**

```python
#Function untuk menghilangkan additional stopwords
def remove_additional_stopwords(text):
    # Additional stopwords yang akan dihilangkan
    words_to_remove = ["yaa", "xquitavee", "ga", "yah", "ayu", "cebokin", "lo", "lu", "gw", "yg", "yang", "dong", "gue", "aplikasi", "premium", "ready", "y", "km", "bran", "full", "gue","juga","dari","admin","atau","tapi","dan","zona uang","tdk","ya","gbk","oleh","pada","cuma","gak","sih","soal","makin","pa","dgn","di","rizky ridho","sama",
                       "dgn","dengan","tsb","dia","klo","aja","kalo","ke","kok","nya","indodax","masa","kok","adalah","tau","apa","klo","itu","banget","si","ada","aku","zonauang","ma","mau","banget","tuh","loklok","youku","cak","jd","saya","tersebut","mulai","harus","dulu","kita","saja","login","warranty","fast","wa","stefanan","notaslim",
                       "mardial","tanmarm","agilafias","masih","lama","terus"]

    # Regex pattern untuk menyesuaikan kalimat
    pattern = r'\b(?:' + '|'.join(words_to_remove) + r')\b'
    return re.sub(pattern, '', text)

# Apply the function to the specified column.
df['full_text'] = df['full_text'].apply(remove_additional_stopwords)
```

**Remove Stopword & Tokenizing**

```python
import nltk

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

import pandas as pd
from nltk.corpus import stopwords

nltk.download('stopwords')

def tokenize_and_remove_stopwords(text):
    tokens = nltk.word_tokenize(text)
    stop_words = set(stopwords.words('indonesian'))
    filtered_tokens = [word for word in tokens if word not in stop_words]
    return filtered_tokens

tokenized_text = 'full_text'
# Apply the function to the 'full_text' column
df['tokenized_text'] = df['full_text'].apply(tokenize_and_remove_stopwords)

print(df[['full_text', 'tokenized_text']])
```

```python
# Inisialisasi SentimentIntensityAnalyzer
sia = SentimentIntensityAnalyzer()

# Fungsi untuk menentukan sentimen berdasarkan skor komponen
def get_sentiment(text):
    # Check if the input is a string
    if isinstance(text, str):
        score = sia.polarity_scores(text)
        if score['compound'] >= 0.05:
            return 'positif'
        elif score['compound'] == 0:
            return 'netral'
        else:
            return 'negatif'
    else:
        # Handle non-string values (e.g., return 'netral' or handle differently)
        return 'netral'

# Terapkan fungsi pelabelan otomatis ke kolom teks
df['sentimen'] = df['full_text'].apply(get_sentiment)  # Sesuaikan 'text' sesuai dengan nama kolom Anda

# Menampilkan data dengan label sentimen
df.head()
```

```python
import matplotlib.pyplot as plt
from wordcloud import WordCloud

def plot_wordcloud(topic_terms, title):
    wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(topic_terms)
    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.title(title)
    plt.axis('off')
    plt.show()

# Visualisasi kata-kata utama dari setiap topik
for i, topic in lda_model.show_topics(formatted=False, num_words=10):
    topic_terms = dict(topic)
    plot_wordcloud(topic_terms, f'Topic {i}')
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Contoh data
data = [
    "gibran", "kaskus", "chillipari", "fufufafa", "menkominfo", "data pribadi", "akun", "prabowo", "diduga", "gibrantweet"
]

# Preprocessing dengan CountVectorizer
vectorizer = CountVectorizer(stop_words='english')
data_vectorized = vectorizer.fit_transform(data)

# Membuat model LDA
lda_model = LatentDirichletAllocation(n_components=3, random_state=42)
lda_model.fit(data_vectorized)

# Mendapatkan daftar kata-kata dari vectorizer
words = vectorizer.get_feature_names_out()

# Membuat dataframe untuk topik dan kata-kata
top_n = 10  # Jumlah kata teratas
topics_df = pd.DataFrame()

for topic_idx, topic in enumerate(lda_model.components_):
    top_words_idx = topic.argsort()[:-top_n - 1:-1]
    top_words = [words[i] for i in top_words_idx]
    top_words_df = pd.DataFrame({
        'term': top_words,
        'beta': topic[top_words_idx],
        'topic': f'Topic {topic_idx + 1}'
    })
    topics_df = pd.concat([topics_df, top_words_df], axis=0)

# Plot menggunakan seaborn
plt.figure(figsize=(10, 7))
sns.barplot(x='beta', y='term', hue='topic', data=topics_df, dodge=False)
plt.title('Top Terms in Each Topic')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer

# Contoh data
data = [
    "gibran", "kaskus", "chillipari", "fufufafa", "menkominfo", "data pribadi",
    "akun", "prabowo", "diduga", "gibrantweet"
]

# Preprocessing dengan CountVectorizer (menghitung frekuensi kata)
vectorizer = CountVectorizer(stop_words='english')  # Stopwords bisa diubah jika diperlukan
data_vectorized = vectorizer.fit_transform(data)

# Mendapatkan daftar kata-kata dan frekuensinya
words = vectorizer.get_feature_names_out()
word_counts = data_vectorized.toarray().sum(axis=0)

# Membuat dataframe dari kata-kata dan frekuensinya
word_freq_df = pd.DataFrame({'term': words, 'frequency': word_counts})

# Mengurutkan berdasarkan frekuensi tertinggi
word_freq_df = word_freq_df.sort_values(by='frequency', ascending=False)

# Plot menggunakan seaborn
plt.figure(figsize=(10, 7))
sns.barplot(x='frequency', y='term', data=word_freq_df, palette='viridis')
plt.title('Frekuensi Kata di Seluruh Data')
plt.xlabel('Frekuensi')
plt.ylabel('Kata')
plt.show()
```
